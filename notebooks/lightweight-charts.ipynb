```python
import sys
import pandas as pd
from lightweight_charts import Chart
from datetime import datetime

project_root = r"D:\Work\DeepVision\nexora-trading-framework"
if project_root not in sys.path:
    sys.path.append(project_root)

from app.visualization.chart_utils import df_to_ligthweigth_chart
from app.tradingview.types import EqualLevel, OrderBlock
from app.rendering.renders.structure_renderer import EqualLevelRender, OrderBlockRender

file = rf'{project_root}\backtests_data\BTCUSDT\2000-01-01_2026-06-01_1d.csv'
```

```python
df = df_to_ligthweigth_chart(fpath=file)
chart = Chart(width=1600, height=900)
chart.set(df)
```

```python
# Render OB
order_block = OrderBlock(top=16291.223, bottom=15599.046, timestamp=datetime(2022, 11, 21), bar_index=2987, direction=1, mitigated=False)
order_block_render = OrderBlockRender(order_block=order_block)
order_block_render.sync(chart)
```

```python
# Render EQL
c1 = df.iloc[2460]
c2 = df.iloc[2500]

equal_level = EqualLevel(c1=c1, c2=c2, first_index=3438, second_index=3443, price=52945.0507, bullish=False, first_time=datetime(2024, 2, 15), second_time=datetime(2024, 2, 20))
equal_level_render = EqualLevelRender(equal_level=equal_level)
equal_level_render.sync(chart)
```
