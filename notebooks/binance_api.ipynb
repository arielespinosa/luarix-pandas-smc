```python
import sys
from datetime import datetime
import pandas_ta_classic as ta

project_root = r"D:\Work\DeepVision\nexora-trading-framework"
if project_root not in sys.path:
    sys.path.append(project_root)

from app.brokers.factory import BrokerFactory, AvailableBrokers
from app.brokers.market_data.provider import DataFormat
from app.core.models.orders import OrderRequest, OrderStatus
from app.core.models.signals import Side
```

```python
SYMBOL = "BTCUSD"
start = datetime(2024, 9, 1)
end = datetime(2026, 6, 26)
interval = "1M"
broker = BrokerFactory.get_broker(broker_name=AvailableBrokers.BINANCE, testnet=False, test_order=True)
```

```python
actives = broker.list_active()
actives
```

```python
monthly_candles = broker.market_data_provider.get_historical_candles(SYMBOL, start=start, end=end, interval=interval, format=DataFormat.OHLCV)
monthly_candles
```

```python
# Account Data
account_data = broker.account_data_provider.get_balance()
account_data["USDT"]
```

```python
# Entry Data

# 1. Creo una orden de entrada
entry = OrderRequest(symbol="ETH", side=Side.BUY, quantity=0.0001)
entry_ack, fills = broker.place_order(entry)
print(entry_ack, "\n\n", fills)
```

```python
# 2. Una vez ejecutada, colocar TP + SL juntos como OCO
if entry_ack.status == OrderStatus.FILLED:
    tp_ack, sl_ack = broker.place_oco_order(
        symbol=SYMBOL,
        side=Side.SELL,           # siempre el lado opuesto a la entrada
        quantity=0.0001,
        take_profit_price=52000,  # tu TP
        stop_price=49000,         # trigger del SL
        stop_limit_price=48900,   # límite real del SL (ligeramente peor que el trigger)
    )
```

```python
# 3. Cancelar una orden
cancel_ack = broker.cancel_order(SYMBOL, entry_ack.client_order_id)
cancel_ack
```
